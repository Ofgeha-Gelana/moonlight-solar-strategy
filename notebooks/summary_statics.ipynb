## Data Quality Check

Checking for outliers

```python
def detect_outliers(dataframe1, column):
    Q1 = dataframe1[column].quantile(0.25)
    Q3 = dataframe1[column].quantile(0.75)
    IQR = Q3 - Q1
    outliers = dataframe1[(dataframe1[column] < (Q1 - 1.5 * IQR)) | (dataframe1[column] > (Q3 + 1.5 * IQR))]
    return outliers
```

```python
sensor_cols = ['ModA', 'ModB', 'WS', 'WSgust']
for col in sensor_cols:
    outliers = detect_outliers(df, col)
    print(f"{col}: {len(outliers)} outliers")
```

 outlier removing function

```python
def remove_outliers(df):
    numeric_columns = df.select_dtypes(include=['int64', 'float64'])
    for column in numeric_columns:
        outliers = detect_outliers(df, column)
        df = df.drop(outliers.index)
    return df
```

```python
df = remove_outliers(df)
df.head()
```

Checking for negative values


```python
negative_cols = ['GHI', 'DNI', 'DHI', 'WS', 'WSgust']
negative_values = df[negative_cols].lt(0).sum()
print(negative_values)
```

Drop rows with negative values in GHI, DNI, and DHI

```python
df = df.drop(df[(df['GHI'] < 0) | (df['DNI'] < 0) | (df['DHI'] < 0)].index)
```

Line Plot of GHI, DNI, DHI, Tamb (Sampled Daily)

## Wind Analysis

Generate Wind Rose Plot

```python
from windrose import WindroseAxes

# Subsample for plotting (optional for performance)
sample = df[['WS', 'WD']].dropna().sample(50000, random_state=42)

# Windrose Plot
ax = WindroseAxes.from_ax()
ax.bar(sample['WD'], sample['WS'], normed=True, opening=0.8, edgecolor='white')
ax.set_legend()
plt.title("Wind Rose: Wind Speed vs Direction")
plt.show()
```

Radial Plot of Wind Direction Variability

```python
# Bin wind direction into 30° sectors
import numpy as np
df['WD_bin'] = pd.cut(df['WD'], bins=np.arange(0, 390, 30), right=False)
wd_var = df.groupby('WD_bin')['WDstdev'].mean()

# Polar plot
theta = np.radians(np.arange(0, 360, 30))
radii = wd_var.values

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
bars = ax.bar(theta, radii, width=np.radians(30), bottom=0.0, alpha=0.6)

ax.set_theta_zero_location('N')
ax.set_theta_direction(-1)
ax.set_title("Wind Direction Variability (WDstdev) by Sector")
plt.show()
```

```python
# Select numerical columns only (ignore 'Cleaning', 'Comments')
num_cols = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'Tamb', 'TModA', 'TModB', 
            'WS', 'WSgust', 'WSstdev', 'WD', 'WDstdev', 'BP', 'RH', 'Precipitation']

from scipy.stats import zscore

# Compute z-scores for selected columns
z_scores = df[num_cols].apply(zscore)

# Flag entries where absolute z-score > threshold (commonly 3)
outlier_mask = (z_scores.abs() > 3)

# Count of outliers per column
outlier_counts = outlier_mask.sum().sort_values(ascending=False)
print("Outlier counts per column:")
print(outlier_counts)
```

## Bubble charts

Bubble Chart (GHI vs Tamb, colored by WS, bubble size = RH)

```python

# Sample data for performance
sampled_df = df[['GHI', 'Tamb', 'WS', 'RH']].dropna().sample(10000, random_state=42)

# Normalize bubble size for RH
rh_norm = (sampled_df['RH'] - sampled_df['RH'].min()) / (sampled_df['RH'].max() - sampled_df['RH'].min()) * 300

plt.figure(figsize=(12, 8))
scatter = plt.scatter(
    sampled_df['GHI'], 
    sampled_df['Tamb'], 
    c=sampled_df['WS'], 
    s=rh_norm, 
    cmap='viridis', 
    alpha=0.6, 
    edgecolor='k'
)

plt.colorbar(scatter, label='Wind Speed (m/s)')
plt.xlabel('GHI (W/m²)')
plt.ylabel('Ambient Temperature (°C)')
plt.title('Bubble Chart: GHI vs Tamb (Bubble Size = RH, Color = WS)')
plt.grid(True)
plt.tight_layout()
plt.show()
```
